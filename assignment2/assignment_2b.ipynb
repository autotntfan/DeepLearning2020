```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

def display_distributions(widths):
  if not widths:
     print('The input list is empty!')
     return
  bin_range = range(0, 201, 10)
  n, bins, patches = plt.hist(x=widths, bins=bin_range, color='#0504aa', alpha=0.7, rwidth=0.85)
  plt.rcParams.update({'font.size': 14})
  plt.grid(axis='y', alpha=0.75)
  plt.xlabel('Width')
  plt.ylabel('Counts')
  maxfreq = n.max()
  plt.ylim(top=np.round(maxfreq / 10) * 10 if maxfreq % 10 else maxfreq + 10)
  plt.show()

if __name__ == '__main__':
    anno_file = 'annotation.txt'
    width=[]
    W10,W20,W30,W40,W50,W50_plus,count=0,0,0,0,0,0,0
    with open(anno_file, 'r') as fp:
        lines = fp.readlines()
  # Enter your code here (20%)
  # You might use the following to access each line
    for line in lines:
        annotation = line.strip().split(' ')
        annotation=[int(i) for i in  annotation[1:]]
        data=np.array(annotation)
        bbox=data.reshape(-1,4)
        for j in bbox:
            count+=1
            w=j[2]-j[0]+1
            if j[0]>=0 and j[1]>=0 and j[2]>j[0] and j[3]>j[1]:
                width.append(w)
            """
            #這邊可以用來看出哪些是不合格的資料
            else:
                print(j)
            """
         
    for length in width:
        if length<10:
            W10+=1
        if length>=10 and length<20:
            W20+=1
        if length>=20 and length<30:
            W30+=1
        if length>=30 and length<40:
            W40+=1
        if length>=40 and length<50:
            W50+=1
        if length>=50:
            W50_plus+=1
    print("""      width < 10: {}\n10 <= width < 20: {}\n20 <= width < 30: {}\n30 <= width < 40: {}\n40 <= width < 50: {}
    50 <= width : {}""".format(W10,W20,W30,W40,W50,W50_plus))
    print("The number of invalid width is:",count-len(width))
            
            
    

  # After you finish Step 2,
  # you may use the following function to visualize the distribution of widths  
  # display_distributions(widths)
```
